# Step 1: Analyze the Original Kaggle Dataset
Before generating synthetic data, inspect the source dataset in Python (using Pandas) to extract its blueprint:
- **Schema:** Check column names, data types (`int`, `float`, `object`, `datetime`). 
- **Distributions & Stats:** Run `.describe()` and `.info()` to see means, standard deviations, min/max values, and null counts.
- **Categories:** Identify unique values and frequencies for categorical columns (e.g., using `df['column'].value_counts(normalize=True)`).
